```python
# Bagging
import numpy as nm
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix
from matplotlib.colors import ListedColormap
from sklearn.metrics import confusion_matrix,accuracy_score
from sklearn import model_selection
from sklearn.ensemble import BaggingClassifier
from sklearn.naive_bayes import GaussianNB
import warnings
warnings.filterwarnings('ignore')
dataset = pd.read_csv("/content/iris.csv")
dataset.head()
```

```python
x=dataset.iloc[:,[0,1,2,3]].values
x
```

```python
y = dataset.iloc[:,4].values
y
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.25, random_state=0)
Single = GaussianNB()
Single.fit(x_train,y_train)
```

```python
y_pred = Single.predict(x_test)
print("Predicted values for single Naive Bayes Classifier")
y_pred
```

```python
Acc_Single=accuracy_score(y_test,y_pred)*100
print("\n\nAccuracy using single Naive Bayes Classifier:",Acc_Single)
```

```python
cm = confusion_matrix(y_test,y_pred)
print("\n\n ConfusionMatrix-using siingle NaiveBayesClassifier:\n",cm)
```

```python
base_cls=GaussianNB()
num_cls = 100
Bag=BaggingClassifier(estimator=base_cls,n_estimators=num_cls,random_state=0)
Bag.fit(x_train,y_train)
```

```python
results = model_selection.cross_val_score(Bag,x_test,y_test,cv = 10)
print("\n\nAccuracy usingBagged set of Naive Bayes Classifier:",results.mean()*100)
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.25, random_state=0)
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
adaboost = AdaBoostClassifier(n_estimators=50,learning_rate=0.2)
adaboost.fit(x_train,y_train)
```

```python
adaboost.score(x_test,y_test)
```

```python
y_pred = adaboost.predict(x_test)
print("Predicted values for AdaBoost Classifier:")
y_pred
```

```python
Acc_adaboost = accuracy_score(y_test,y_pred)*100
print("\n\nTest Accuracy using AdaBoost Classisfier:",Acc_adaboost)
```

```python
fig,ax = plt.subplots(figsize=(6,6))
ax.imshow(cm)
ax.grid(False)
ax.xaxis.set(ticks=(0,1,2),ticklabels=('Predicted Setosa','Predicted Versicolor','Predicted Virginica'))
ax.yaxis.set(ticks=(0,1,2),ticklabels=('Actual Setosa', 'Actual Versicolor','Actual Virginica'))
ax.set_ylim(2.5,-0.5)
for i in range(3):
  for j in range(3):
    ax.text(j,i,cm[i,j],ha='center',va='center',color='white')
plt.show()
```
